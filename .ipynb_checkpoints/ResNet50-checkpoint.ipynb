```python
import os
import torch
from torch.utils.data import Dataset
from PIL import Image
import numpy as np
from tqdm import tqdm
from albumentations import Compose, Resize, Normalize
from albumentations.pytorch import ToTensorV2
from torchvision.models.segmentation import deeplabv3_resnet50, DeepLabV3_ResNet50_Weights
import torch.nn as nn
from torch.utils.data import DataLoader
import torch.optim as optim
import time
from sklearn.metrics import precision_score, recall_score
import matplotlib.pyplot as plt
```

```python
#Load dữ liệu bằng Dataset
#Tạo LaRSDataset để load ảnh RGB và mask
class LaRSDataset(Dataset):
    def __init__(self, root_dir, split='train', transform=None):
        self.transform = transform

        # Đường dẫn chính xác
        self.image_dir = os.path.join(root_dir, 'images', split, 'images')
        self.mask_dir = os.path.join(root_dir, 'annotations', 'semantic', split, 'semantic_masks')

        # Lọc file .jpg trong tập train
        self.image_files = sorted([
            f for f in os.listdir(self.image_dir) if f.endswith('.jpg')
        ])
    
    def __len__(self):
        return len(self.image_files)
    
    def __getitem__(self, idx):
        img_name = self.image_files[idx]
        image_path = os.path.join(self.image_dir, img_name)
        mask_name = img_name.replace('.jpg', '.png')
        mask_path = os.path.join(self.mask_dir, mask_name)

        image = np.array(Image.open(image_path).convert('RGB'))
        mask = np.array(Image.open(mask_path)).astype(np.uint8)

        # CHẶN lỗi: gán 255 cho mọi pixel có giá trị > 2
        mask[mask > 2] = 255

        if self.transform:
            transformed = self.transform(image=image, mask=mask)
            image = transformed['image']
            mask = transformed['mask']

        return image, mask.long()
```

```python
# Tạo DataLoader cho tập train và val
#from torch.utils.data import DataLoader
#from albumentations import Compose, Resize, Normalize
#from albumentations.pytorch import ToTensorV2

# (512, 512) -> (256, 256)
# Cấu hình transform dùng chung
transform = Compose([
    Resize(320, 320),
    Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
    ToTensorV2()
])

# Dataset
train_dataset = LaRSDataset("data/lars", split="train", transform=transform)
val_dataset   = LaRSDataset("data/lars", split="val", transform=transform)

# DataLoader
train_loader = DataLoader(train_dataset, batch_size=4, shuffle=True, num_workers=0, drop_last=True)  # Giảm batch_size từ 4 -> 2
val_loader   = DataLoader(val_dataset, batch_size=4, shuffle=False, num_workers=0, drop_last=True)   # Giảm batch_size từ 4 -> 2
```

```python
# Định nghĩa model DeepLabV3+ (ResNet50)
#from torchvision.models.segmentation import deeplabv3_resnet50, DeepLabV3_ResNet50_Weights
#import torch.nn as nn
#import torch

def get_deeplab_model(num_classes=3):
    # Sử dụng weights chuẩn (COCO/VOC) theo khuyến nghị
    weights = DeepLabV3_ResNet50_Weights.DEFAULT

    # Load mô hình với backbone đã pretrained
    model = deeplabv3_resnet50(weights=weights)

    # Thay lớp phân lớp cuối cùng để phù hợp số lớp của bạn
    model.classifier[4] = nn.Conv2d(256, num_classes, kernel_size=1)

    return model.to('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
# Kiểm tra model
model = get_deeplab_model(num_classes=3)
#print(model)
```

```python
# Bật Debug CUDA
#import os
os.environ["CUDA_LAUNCH_BLOCKING"] = "1"
# Training Loop
#import torch
#import torch.nn as nn
#import torch.optim as optim
#from tqdm import tqdm

# ⚡ Tăng tốc cho ảnh có cùng kích thước
torch.backends.cudnn.benchmark = True

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = get_deeplab_model(num_classes=3).to(device)

criterion = nn.CrossEntropyLoss(ignore_index=255)
optimizer = optim.Adam(model.parameters(), lr=1e-4)

num_epochs = 20

for epoch in range(num_epochs):
    model.train()
    total_loss = 0

    for images, masks in tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs}"):
        images, masks = images.to(device), masks.to(device)

        optimizer.zero_grad()
        outputs = model(images)['out']
        loss = criterion(outputs, masks)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch {epoch+1}, Loss: {total_loss:.4f}")
```

```python
#import time
#import numpy as np
#from sklearn.metrics import precision_score, recall_score

def evaluate_full_metrics(model, dataset, num_classes=3, max_fps=10, device='cuda', num_samples=200):
    model.eval()
    loader = torch.utils.data.DataLoader(dataset, batch_size=1, shuffle=False)

    # ---------- BẮT ĐẦU TÍNH FPS ----------
    start = time.time()
    with torch.no_grad():
        for i, (image, _) in enumerate(loader):
            if i >= num_samples: break
            _ = model(image.to(device))['out']
    end = time.time()
    fps = num_samples / (end - start)
    norm_fps = min(fps, max_fps) / max_fps

    # ---------- BẮT ĐẦU TÍNH F1 ----------
    y_true_all = []
    y_pred_all = []
    with torch.no_grad():
        for i, (image, mask) in enumerate(loader):
            if i >= num_samples: break
            image, mask = image.to(device), mask.to(device)
            output = model(image)['out']
            pred = torch.argmax(output, dim=1)

            valid = (mask != 255).view(-1)
            y_true = mask.view(-1)[valid].cpu().numpy()
            y_pred = pred.view(-1)[valid].cpu().numpy()

            y_true_all.extend(y_true)
            y_pred_all.extend(y_pred)

    precision = precision_score(y_true_all, y_pred_all, average='macro', zero_division=0)
    recall    = recall_score(y_true_all, y_pred_all, average='macro', zero_division=0)
    f1 = 2 * precision * recall / (precision + recall + 1e-8)

    # ---------- BẮT ĐẦU TÍNH mIoU ----------
    hist = np.zeros((num_classes, num_classes), dtype=np.int64)
    with torch.no_grad():
        for i, (image, mask) in enumerate(loader):
            if i >= num_samples: break
            image, mask = image.to(device), mask.to(device)
            output = model(image)['out']
            pred = torch.argmax(output, dim=1).squeeze(0)
            mask = mask.squeeze(0)

            valid = (mask != 255)
            pred = pred[valid]
            mask = mask[valid]

            for t, p in zip(mask.view(-1), pred.view(-1)):
                if t < num_classes:
                    hist[t.long(), p.long()] += 1

    iou_per_class = np.diag(hist) / (hist.sum(1) + hist.sum(0) - np.diag(hist) + 1e-8)
    miou = np.mean(iou_per_class)

    # ---------- TÍNH METRIC CUỐI ----------
    metric = 2 * norm_fps * f1 / (norm_fps + f1 + 1e-8)

    # ---------- IN KẾT QUẢ ----------
    print(f"✅ FPS: {fps:.2f}")
    print(f"✅ NormFPS: {norm_fps:.4f}")
    print(f"✅ F1 Score: {f1:.4f}")
    print(f"✅ mIoU: {miou:.4f}")
    print(f"🎯 Final Metric: {metric:.4f}")

    return {
        "fps": fps,
        "norm_fps": norm_fps,
        "f1": f1,
        "miou": miou,
        "metric": metric
    }
```

```python
evaluate_full_metrics(model, val_dataset, num_classes=3)
```

```python
#import matplotlib.pyplot as plt
#import numpy as np
#import torch

def visualize_prediction(model, dataset, idx=0, device='cuda', class_colors=None):
    model.eval()
    image, mask = dataset[idx]
    image_tensor = image.unsqueeze(0).to(device)

    with torch.no_grad():
        output = model(image_tensor)['out']
        pred_mask = torch.argmax(output, dim=1).squeeze(0).cpu().numpy()

    # Convert image tensor to numpy
    image_np = image.permute(1, 2, 0).cpu().numpy()  # [H, W, 3]
    image_np = np.clip(image_np * 255, 0, 255).astype(np.uint8)

    gt_mask = mask.cpu().numpy()

    # Use default color palette if not provided
    if class_colors is None:
        class_colors = {
            0: (247, 195, 37),         #
            1: (41, 167, 224),       #
            2: (90, 75, 164),       #
            255: (127, 127, 127)  # gray for ignored
        }

    def colorize(mask):
        color_mask = np.zeros((mask.shape[0], mask.shape[1], 3), dtype=np.uint8)
        for k, v in class_colors.items():
            color_mask[mask == k] = v
        return color_mask

    plt.figure(figsize=(12, 4))
    plt.subplot(1, 3, 1)
    plt.imshow(image_np)
    plt.title("Ảnh Gốc")

    plt.subplot(1, 3, 2)
    plt.imshow(colorize(gt_mask))
    plt.title("Mask Ground Truth")

    plt.subplot(1, 3, 3)
    plt.imshow(colorize(pred_mask))
    plt.title("Mask Dự đoán")

    plt.tight_layout()
    plt.show()
```

```python
visualize_prediction(model, val_dataset, idx=34)  # Hoặc idx khác
```

```python
import os
import numpy as np
from PIL import Image
import torchvision.transforms as T
import tensorflow as tf

def representative_data_gen():
    image_dir = "data/lars/images/val/images"
    image_files = [os.path.join(image_dir, f) for f in os.listdir(image_dir) if f.endswith(".jpg")]

    preprocess = T.Compose([
        T.Resize((320, 320)),
        T.ToTensor(),
        T.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225))
    ])

    for image_path in image_files[:100]:  # lấy 100 ảnh để làm mẫu
        img = Image.open(image_path).convert("RGB")
        img = preprocess(img)
        img = img.numpy()
        img = np.expand_dims(img, axis=0)
        img = img.astype(np.float32)
        yield [img]
```

```python
# Cell 4: Chuyển từ SavedModel sang TFLite int8
converter = tf.lite.TFLiteConverter.from_saved_model("model_saved")
converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.representative_dataset = representative_data_gen
converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
converter.inference_input_type = tf.uint8
converter.inference_output_type = tf.uint8

tflite_model = converter.convert()

with open("resnet50_int8.tflite", "wb") as f:
    f.write(tflite_model)

print("✅ Đã tạo resnet50_int8.tflite")
```
