```python
# ✅ Tối ưu và sửa mã dựa trên notebook "AI - Demo.ipynb"
# Đã cập nhật: dùng MobileNetV2 thay cho ResNet50, thêm hỗ trợ Quantization Aware Training (QAT), lưu model tốt nhất và log loss/lr từng epoch

import os
import random
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import torch.backends.cudnn
from torch.utils.data import Dataset, DataLoader
from torchvision.models.segmentation import deeplabv3_mobilenet_v3_large, DeepLabV3_MobileNet_V3_Large_Weights
from torchvision.models.quantization import mobilenet_v2
from albumentations import Compose, Resize, Normalize
from albumentations.pytorch import ToTensorV2
from PIL import Image
from tqdm import tqdm
import matplotlib.pyplot as plt
from sklearn.metrics import precision_score, recall_score
import torch.onnx
import onnx
```

```python
# ✅ Dataset xử lý Semantic Segmentation từ LaRS
class LaRSDataset(Dataset):
    def __init__(self, root_dir, split='train', transform=None):
        self.image_dir = os.path.join(root_dir, 'images', split, 'images')
        self.mask_dir = os.path.join(root_dir, 'annotations', 'semantic', split, 'semantic_masks')
        self.image_files = sorted([f for f in os.listdir(self.image_dir) if f.endswith('.jpg')])
        self.transform = transform

    def __len__(self):
        return len(self.image_files)

    def __getitem__(self, idx):
        img_name = self.image_files[idx]
        image_path = os.path.join(self.image_dir, img_name)
        mask_path = os.path.join(self.mask_dir, img_name.replace('.jpg', '.png'))

        image = np.array(Image.open(image_path).convert('RGB'))
        mask = np.array(Image.open(mask_path)).astype(np.uint8)
        mask[mask > 2] = 255

        if self.transform:
            transformed = self.transform(image=image, mask=mask)
            image = transformed['image']
            mask = transformed['mask']

        return image, mask.long()
```

```python
# ✅ Model DeepLabV3+ với MobileNetV2 backbone (QAT-compatible)
def get_deeplab_model(num_classes=3, quantize=False):
    weights = DeepLabV3_MobileNet_V3_Large_Weights.DEFAULT
    model = deeplabv3_mobilenet_v3_large(weights=weights)
    model.classifier[4] = nn.Conv2d(256, num_classes, kernel_size=1)

    if quantize:
        model.qconfig = torch.quantization.get_default_qat_qconfig('fbgemm')
        torch.quantization.prepare_qat(model, inplace=True)

    return model.to('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
# Cấu hình transform dùng chung
transform = Compose([
    Resize(320, 320),
    Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
    ToTensorV2()
])

# Dataset
train_dataset = LaRSDataset("data/lars", split="train", transform=transform)
val_dataset   = LaRSDataset("data/lars", split="val", transform=transform)

# DataLoader
train_loader = DataLoader(train_dataset, batch_size=8, shuffle=True, num_workers=0, pin_memory=True, drop_last=True)
val_loader   = DataLoader(val_dataset, batch_size=8, shuffle=False, num_workers=0, drop_last=True)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = get_deeplab_model(num_classes=3).to(device)
```

```python
from torch.optim import AdamW
from torch.optim.lr_scheduler import StepLR
```

```python
def train_model(model, train_loader, criterion, optimizer, device, num_epochs=30, save_path="best_model.pth"): 
    torch.backends.cudnn.benchmark = True
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.5)
    best_loss = float('inf')
    log = []

    for epoch in range(num_epochs):
        model.train()
        total_loss = 0
        for images, masks in tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs}"):
            images, masks = images.to(device), masks.to(device)
            optimizer.zero_grad()
            outputs = model(images)['out']
            loss = criterion(outputs, masks)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()

        avg_loss = total_loss / len(train_loader)
        lr = optimizer.param_groups[0]['lr']
        log.append((epoch+1, avg_loss, lr))
        print(f"✅ Epoch {epoch+1} | Loss: {avg_loss:.4f} | LR: {lr:.6f}")

        if avg_loss < best_loss:
            best_loss = avg_loss
            torch.save(model.state_dict(), save_path)
            print(f"📌 Saved best model to {save_path}")

        scheduler.step()

    with open("train_log.csv", "w") as f:
        f.write("epoch,loss,lr\n")
        for e, l, r in log:
            f.write(f"{e},{l},{r}\n")
    print("📊 Training log saved to train_log.csv")
```

```python
criterion = nn.CrossEntropyLoss(ignore_index=255)
optimizer = optim.Adam(model.parameters(), lr=1e-4)

train_model(model, train_loader, criterion, optimizer, device, num_epochs=30)
```

```python
def evaluate_full_metrics(model, dataset, num_classes=3, max_fps=10, device='cuda', num_samples=200):
    model.eval()
    loader = DataLoader(dataset, batch_size=1, shuffle=False)

    import time
    start = time.time()
    with torch.no_grad():
        for i, (image, _) in enumerate(loader):
            if i >= num_samples: break
            _ = model(image.to(device))['out']
    fps = num_samples / (time.time() - start)
    norm_fps = min(fps, max_fps) / max_fps

    y_true, y_pred = [], []
    with torch.no_grad():
        for i, (image, mask) in enumerate(loader):
            if i >= num_samples: break
            out = model(image.to(device))['out']
            pred = torch.argmax(out, dim=1).cpu().view(-1)
            mask = mask.view(-1)
            valid = (mask != 255)
            y_true.extend(mask[valid].numpy())
            y_pred.extend(pred[valid].numpy())
    p = precision_score(y_true, y_pred, average='macro', zero_division=0)
    r = recall_score(y_true, y_pred, average='macro', zero_division=0)
    f1 = 2 * p * r / (p + r + 1e-8)

    hist = np.zeros((num_classes, num_classes), dtype=np.int64)
    with torch.no_grad():
        for i, (image, mask) in enumerate(loader):
            if i >= num_samples: break
            out = model(image.to(device))['out']
            pred = torch.argmax(out, dim=1).cpu().squeeze()
            mask = mask.squeeze()
            valid = (mask != 255)
            for t, p in zip(mask[valid].view(-1), pred[valid].view(-1)):
                if t < num_classes:
                    hist[t, p] += 1
    iou = np.diag(hist) / (hist.sum(1) + hist.sum(0) - np.diag(hist) + 1e-8)
    miou = np.mean(iou)

    metric = 2 * norm_fps * f1 / (norm_fps + f1 + 1e-8)
    # -------IN KẾT QUẢ--------
    print(f"-     FPS: {fps:.2f}")
    print(f"- NormFPS: {norm_fps:.4f}")
    print(f"-      F1: {f1:.4f}")
    print(f"-    mIoU: {miou:.4f}")
    print(f"-  Metric: {metric:.4f}")
    return {"fps": fps, "f1": f1, "miou": miou, "metric": metric}
```

```python
evaluate_full_metrics(model, val_dataset, num_classes=3, device=device)
```

```python
def visualize_prediction(model, dataset, idx=0, device='cuda'):
    model.eval()
    image, mask = dataset[idx]
    image_tensor = image.unsqueeze(0).to(device)
    with torch.no_grad():
        output = model(image_tensor)['out']
        pred = torch.argmax(output, dim=1).squeeze().cpu().numpy()

    # ✅ Unnormalize ảnh
    img = image.permute(1, 2, 0).cpu().numpy()
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    img = (img * std + mean) * 255
    img = img.astype(np.uint8)
    
    gt = mask.cpu().numpy()

    def colorize(mask):
        colors = {
            0: (247, 195, 37),    # obstacle
            1: (41, 167, 224),    # water
            2: (90, 75, 164),     # sky
            255: (127, 127, 127)  # gray for ignored
        }
        out = np.zeros((*mask.shape, 3), dtype=np.uint8)
        for k, v in colors.items():
            out[mask == k] = v
        return out

    plt.figure(figsize=(12, 4))
    plt.subplot(1, 3, 1); plt.imshow(img.astype(np.uint8)); plt.title("Original Photo")
    plt.subplot(1, 3, 2); plt.imshow(colorize(gt)); plt.title("Ground Truth")
    plt.subplot(1, 3, 3); plt.imshow(colorize(pred)); plt.title("Predict")
    plt.show()
```

```python
visualize_prediction(model, val_dataset, idx=4)
```

```python
# ✅ Export mô hình sang TorchScript (float16 optional)
def export_to_torchscript(pth_path="best_model.pth", output_path="model.pt", input_size=(1, 3, 320, 320), num_classes=3, use_fp16=True):
    model = get_deeplab_model(num_classes=num_classes)
    model.load_state_dict(torch.load(pth_path, map_location=torch.device('cpu')))
    model.eval()

    dummy_input = torch.randn(*input_size)
    if use_fp16:
        model = model.half()
        dummy_input = dummy_input.half()

    model.cpu()  # Ensure model and input are on the same device
    output = model(dummy_input)
    if isinstance(output, dict):
        output = output["out"]
    traced_model = torch.jit.trace(model, dummy_input, strict=False)
    torch.jit.save(traced_model, output_path)
    print(f"✅ Saved TorchScript model to: {output_path} (float16={use_fp16})")
```

```python
export_to_torchscript("best_model.pth", "model.pt", use_fp16=True)
```

```python
# ✅ Export mô hình sang ONNX (không dùng onnx-tf)
def export_to_onnx(model, input_size=(1, 3, 320, 320), onnx_path="model.onnx"):
    model.eval()
    device = next(model.parameters()).device
    dummy_input = torch.randn(*input_size).to(device)
    torch.onnx.export(model, dummy_input, onnx_path,
                      input_names=['input'], output_names=['output'],
                      opset_version=11)
    print(f"✅ Exported ONNX model to {onnx_path}")
```

```python
export_to_onnx(model, input_size=(1, 3, 320, 320))
```

```python
def convert_onnx_to_tflite(onnx_path="model.onnx", saved_model_dir="saved_model", tflite_path="model_float16.tflite", quantized=False, quant_type="float16"):
    import shutil
    import tensorflow as tf
    import onnx
    from onnx_tf.backend import prepare

    # Load ONNX
    onnx_model = onnx.load(onnx_path)
    tf_rep = prepare(onnx_model)

    if os.path.exists(saved_model_dir):
        shutil.rmtree(saved_model_dir)
    tf_rep.export_graph(saved_model_dir)
    print(f"✅ ONNX → SavedModel: {saved_model_dir}")

    # Convert to TFLite
    converter = tf.lite.TFLiteConverter.from_saved_model(saved_model_dir)

    # SAFETY: Tắt converter mới nếu lỗi
    converter.experimental_new_converter = False

    if quantized:
        converter.optimizations = [tf.lite.Optimize.DEFAULT]
        if quant_type == "float16":
            converter.target_spec.supported_types = [tf.float16]
            print("⚙️ Float16 quantization bật")
        elif quant_type == "int8":
            import numpy as np
            converter.representative_dataset = lambda: [(np.random.rand(1, 320, 320, 3).astype(np.float32),)]
            converter.inference_input_type = tf.uint8
            converter.inference_output_type = tf.uint8
            print("⚙️ Int8 quantization bật")

    try:
        tflite_model = converter.convert()
        with open(tflite_path, "wb") as f:
            f.write(tflite_model)
        print(f"✅ TFLite exported: {tflite_path}")
    except Exception as e:
        print(f"❌ Lỗi khi convert TFLite: {e}")
```

```python
# convert_onnx_to_tflite(quantized=True, quant_type="float16")
convert_onnx_to_tflite(quantized=True, quant_type="float16")
```

```python
# ✅ Load và test file .tflite đơn giản

def load_tflite_model(tflite_path="model_float32.tflite"):
    import tensorflow as tf
    interpreter = tf.lite.Interpreter(model_path=tflite_path)
    interpreter.allocate_tensors()
    input_details = interpreter.get_input_details()
    output_details = interpreter.get_output_details()

    dummy_input = np.random.rand(1, 3, 320, 320).astype(np.float32)
    interpreter.set_tensor(input_details[0]['index'], dummy_input)
    interpreter.invoke()
    output = interpreter.get_tensor(output_details[0]['index'])
    print("✅ TFLite model chạy thành công. Output shape:", output.shape)
    return output
```

```python
load_tflite_model("model_float32.tflite")
```

```python
# ✅ Test TFLite model trên ảnh thực tế từ val_dataset

def test_tflite_on_sample(tflite_path, dataset, idx=0):
    import tensorflow as tf
    interpreter = tf.lite.Interpreter(model_path=tflite_path)
    interpreter.allocate_tensors()
    input_details = interpreter.get_input_details()
    output_details = interpreter.get_output_details()

    image, _ = dataset[idx]  # Bỏ mask
    image_np = image.numpy()  # PyTorch Tensor: (3, 320, 320)
    image_np = np.expand_dims(image_np, axis=0).astype(np.float32)  # (1, 3, 320, 320)

    interpreter.set_tensor(input_details[0]['index'], image_np)
    interpreter.invoke()
    output = interpreter.get_tensor(output_details[0]['index'])
    print("✅ Dự đoán từ ảnh val_dataset. Output shape:", output.shape)
    return output
```

```python
test_tflite_on_sample("model_float32.tflite", val_dataset, idx=4)
```

```python
import tensorflow as tf
def inspect_tflite_model(model_path):
    interpreter = tf.lite.Interpreter(model_path=model_path)
    interpreter.allocate_tensors()
    for detail in interpreter.get_input_details():
        print("📥 Input:", detail)
    for detail in interpreter.get_output_details():
        print("📤 Output:", detail)
```
