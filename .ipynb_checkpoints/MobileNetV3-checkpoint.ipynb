```python
import os
import torch
import torch.nn as nn
from torchvision.models.segmentation import deeplabv3_mobilenet_v3_large
from torchvision.models._utils import IntermediateLayerGetter
from albumentations import Compose, Resize, Normalize, RandomCrop
from albumentations.pytorch import ToTensorV2
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import onnx
import onnxruntime
import tensorflow as tf
from onnx_tf.backend import prepare
```

```python
class LaRSDataset(torch.utils.data.Dataset):
    def __init__(self, root_dir, split='train', transform=None, img_size=(320, 320)):
        self.img_dir = os.path.join(root_dir, 'images', split)
        self.mask_dir = os.path.join(root_dir, 'annotations', 'semantic', split)
        self.img_files = sorted([f for f in os.listdir(self.img_dir) if f.endswith('.jpg')])
        self.transform = transform
        self.img_size = img_size

    def __getitem__(self, idx):
        img_path = os.path.join(self.img_dir, self.img_files[idx])
        mask_path = os.path.join(self.mask_dir, self.img_files[idx].replace('.jpg', '.png'))
        
        # Tải ảnh và mask với lazy loading
        img = np.array(Image.open(img_path).convert('RGB'))
        mask = np.array(Image.open(mask_path))
        mask = np.where(mask > 2, 255, mask)  # Xử lý lớp background

        if self.transform:
            augmented = self.transform(image=img, mask=mask)
            img, mask = augmented['image'], augmented['mask']
        
        return img, mask.long()

    def __len__(self):
        return len(self.img_files)
```

```python
def get_mobilenetv3_model(num_classes=3, pretrained=True):
    # Load model với aux_head=False để giảm kích thước
    model = deeplabv3_mobilenet_v3_large(
        pretrained=pretrained,
        num_classes=num_classes,
        aux_loss=False
    )
    
    # Tối ưu backbone cho segmentation
    backbone = model.backbone
    backbone = IntermediateLayerGetter(backbone, return_layers={
        'high_level_features': 'out',
        'low_level_features': 'low_level'
    })
    model.backbone = backbone
    
    # Giảm channels cho ASPP module
    model.classifier[0].project[0] = nn.Conv2d(256, 128, kernel_size=1)  # Giảm từ 256->128 channels
    
    return model
```

```python
def get_dataloaders(data_dir, batch_size=8, img_size=(320, 320)):
    train_transform = Compose([
        RandomCrop(img_size[0], img_size[1], p=1.0),
        Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
        ToTensorV2()
    ])
    
    val_transform = Compose([
        Resize(img_size[0], img_size[1]),
        Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
        ToTensorV2()
    ])

    train_dataset = LaRSDataset(data_dir, split='train', transform=train_transform)
    val_dataset = LaRSDataset(data_dir, split='val', transform=val_transform)
    
    train_loader = torch.utils.data.DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        pin_memory=True,
        num_workers=2,
        drop_last=True
    )
    
    val_loader = torch.utils.data.DataLoader(
        val_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=2
    )
    
    return train_loader, val_loader
```

```python
def train_model(model, train_loader, val_loader, epochs=30, lr=1e-4):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model.to(device)
    
    criterion = nn.CrossEntropyLoss(ignore_index=255)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-5)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'min', patience=3)
    
    best_miou = 0.0
    
    for epoch in range(epochs):
        model.train()
        epoch_loss = 0.0
        
        for images, masks in train_loader:
            images, masks = images.to(device), masks.to(device)
            
            optimizer.zero_grad(set_to_none=True)  # Tối ưu bộ nhớ
            
            outputs = model(images)['out']
            loss = criterion(outputs, masks)
            loss.backward()
            optimizer.step()
            
            epoch_loss += loss.item()
        
        # Đánh giá trên validation set
        val_miou = evaluate(model, val_loader, device)
        scheduler.step(val_miou)
        
        print(f'Epoch {epoch+1}/{epochs} | Loss: {epoch_loss/len(train_loader):.4f} | mIoU: {val_miou:.4f}')
        
        # Lưu model tốt nhất
        if val_miou > best_miou:
            best_miou = val_miou
            torch.save(model.state_dict(), f'best_model_mobilenetv3.pth')
    
    return model
```

```python
def export_to_onnx(model, output_path="model.onnx", img_size=(320, 320)):
    dummy_input = torch.randn(1, 3, img_size[0], img_size[1])
    
    torch.onnx.export(
        model,
        dummy_input,
        output_path,
        input_names=['input'],
        output_names=['output'],
        dynamic_axes={
            'input': {0: 'batch_size'},
            'output': {0: 'batch_size'}
        },
        opset_version=13,
        do_constant_folding=True
    )
    
    print(f"✅ ONNX model exported to {output_path}")
```

```python
def convert_to_tflite_int8(onnx_path, calibration_loader, output_path="model_quant.tflite"):
    # 1. ONNX → TensorFlow
    onnx_model = onnx.load(onnx_path)
    tf_rep = prepare(onnx_model)
    tf_rep.export_graph("tf_saved_model")
    
    # 2. Tạo representative dataset từ validation loader
    def representative_dataset():
        for images, _ in calibration_loader:
            images = (images.numpy() * 255).astype(np.uint8)  # Scale [0,255]
            for img in images:
                yield [np.expand_dims(img, axis=0)]
    
    # 3. Cấu hình converter
    converter = tf.lite.TFLiteConverter.from_saved_model("tf_saved_model")
    converter.optimizations = [tf.lite.Optimize.DEFAULT]
    converter.representative_dataset = representative_dataset
    converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
    converter.inference_input_type = tf.uint8
    converter.inference_output_type = tf.uint8
    
    # 4. Convert và lưu
    tflite_model = converter.convert()
    with open(output_path, "wb") as f:
        f.write(tflite_model)
    
    print(f"✅ TFLite INT8 model saved to {output_path}")
```

```python
def evaluate(model, data_loader, device):
    model.eval()
    total_iou = 0.0
    
    with torch.no_grad():
        for images, masks in data_loader:
            images, masks = images.to(device), masks.to(device)
            outputs = model(images)['out']
            preds = torch.argmax(outputs, dim=1)
            
            # Tính IoU đơn giản (có thể thay bằng hàm chính xác hơn)
            intersection = (preds == masks).sum()
            union = (preds | masks).sum()
            total_iou += (intersection / (union + 1e-10)).item()
    
    return total_iou / len(data_loader)

def test_tflite_model(tflite_path, test_image):
    interpreter = tf.lite.Interpreter(model_path=tflite_path)
    interpreter.allocate_tensors()
    
    input_details = interpreter.get_input_details()
    output_details = interpreter.get_output_details()
    
    # Chuẩn bị input (scale 0-255)
    input_data = (test_image * 255).astype(np.uint8)
    input_data = np.expand_dims(input_data, axis=0)
    
    interpreter.set_tensor(input_details[0]['index'], input_data)
    interpreter.invoke()
    output = interpreter.get_tensor(output_details[0]['index'])
    
    return output
```

```python
# 1. Khởi tạo model và data loader
model = get_mobilenetv3_model(num_classes=3)
train_loader, val_loader = get_dataloaders("data/lars", batch_size=8)
```

```python
# 2. Huấn luyện (không QAT)
model = train_model(model, train_loader, val_loader, epochs=30)
```

```python
# 3. Xuất ONNX
export_to_onnx(model, "model.onnx")
```

```python
# 4. Chuyển sang TFLite int8 (Post-training Quantization)
convert_to_tflite_int8("model.onnx", val_loader, "model_quant_int8.tflite")
```
