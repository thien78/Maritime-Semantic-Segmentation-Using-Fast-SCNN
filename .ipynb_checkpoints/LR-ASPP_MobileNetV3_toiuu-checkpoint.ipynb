```python
# CELL1: Import các thư viện cần thiết
import os
import numpy as np
from PIL import Image
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchvision.models.segmentation import lraspp_mobilenet_v3_large
import albumentations as A
from albumentations.pytorch import ToTensorV2
from albumentations import Compose, Resize, Normalize
import matplotlib.pyplot as plt
from sklearn.metrics import precision_score, recall_score
from tqdm import tqdm
from torch.cuda.amp import autocast, GradScaler  # Mixed precision training
```

```python
# CELL2: Dataset class
class LaRSDataset(Dataset):
    def __init__(self, root_dir, split='train', transform=None):
        self.image_dir = os.path.join(root_dir, 'images', split, 'images')
        self.mask_dir = os.path.join(root_dir, 'annotations', 'semantic', split, 'semantic_masks')
        self.image_files = sorted([f for f in os.listdir(self.image_dir) if f.endswith('.jpg')])
        self.transform = transform

    def __len__(self):
        return len(self.image_files)

    def __getitem__(self, idx):
        img_name = self.image_files[idx]
        image_path = os.path.join(self.image_dir, img_name)
        mask_path = os.path.join(self.mask_dir, img_name.replace('.jpg', '.png'))

        image = np.array(Image.open(image_path).convert('RGB'))
        mask = np.array(Image.open(mask_path)).astype(np.uint8)
        mask[mask > 2] = 255  # Set any class > 2 to ignore index (255)

        if self.transform:
            transformed = self.transform(image=image, mask=mask)
            image = transformed['image']
            mask = transformed['mask']

        return image, mask.long()
```

```python
# CELL3: Data preparation
# Cấu hình transform dùng chung
transform = Compose([
    Resize(320, 320),
    Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
    ToTensorV2()
])

# Dataset
train_dataset = LaRSDataset("data/lars", split="train", transform=transform)
val_dataset = LaRSDataset("data/lars", split="val", transform=transform)

# DataLoader
train_loader = DataLoader(train_dataset, batch_size=4, shuffle=True, num_workers=0, drop_last=True) 
val_loader = DataLoader(val_dataset, batch_size=4, shuffle=False, num_workers=0, drop_last=True)
```

```python
# CELL4: Model definition
def get_model(num_classes=3):
    # Load pretrained MobileNetV3 with LR-ASPP
    model = lraspp_mobilenet_v3_large(pretrained=True)
    
    # Modify the classifier heads for our number of classes
    # Low-level features classifier
    model.classifier.low_classifier = nn.Conv2d(40, num_classes, kernel_size=1)
    # High-level features classifier
    model.classifier.high_classifier = nn.Conv2d(128, num_classes, kernel_size=1)
    
    return model
```

```python
def train_model(model, train_loader, val_loader, num_epochs=30, device='cuda'):
    # Khởi tạo model và các thành phần huấn luyện
    model = model.to(device)
    
    # Sử dụng AdamW với weight decay
    optimizer = optim.AdamW(model.parameters(), lr=0.001, weight_decay=1e-4)
    
    # Scheduler cosine với warmup
    scheduler = optim.lr_scheduler.OneCycleLR(
        optimizer,
        max_lr=0.01,
        epochs=num_epochs,
        steps_per_epoch=len(train_loader),
        pct_start=0.1
    )
    
    criterion = nn.CrossEntropyLoss(ignore_index=255)
    
    # Khởi tạo gradient scaler cho mixed precision
    scaler = torch.cuda.amp.GradScaler()
    
    # Theo dõi các metrics
    best_val_loss = float('inf')
    best_miou = 0.0
    early_stop_counter = 0
    patience = 5
    
    train_losses = []
    val_losses = []
    val_mious = []
    
    for epoch in range(num_epochs):
        model.train()
        epoch_train_loss = 0.0
        
        # Training loop với progress bar
        for images, masks in tqdm(train_loader, desc=f'Epoch {epoch+1}/{num_epochs}'):
            images = images.to(device, non_blocking=True)
            masks = masks.to(device, non_blocking=True)
            
            # Zero gradient
            optimizer.zero_grad(set_to_none=True)  # Tối ưu memory
            
            # Forward pass với mixed precision
            with torch.cuda.amp.autocast():
                outputs = model(images)['out']
                loss = criterion(outputs, masks)
            
            # Backward pass với gradient scaling
            scaler.scale(loss).backward()
            
            # Gradient clipping
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            
            # Cập nhật weights
            scaler.step(optimizer)
            scaler.update()
            
            # Cập nhật learning rate
            scheduler.step()
            
            epoch_train_loss += loss.item()
        
        # Tính toán average training loss
        avg_train_loss = epoch_train_loss / len(train_loader)
        train_losses.append(avg_train_loss)
        
        # Validation phase
        model.eval()
        val_loss = 0.0
        total_pixels = 0
        correct_pixels = 0
        
        with torch.no_grad():
            for images, masks in val_loader:
                images = images.to(device, non_blocking=True)
                masks = masks.to(device, non_blocking=True)
                
                # Forward pass với mixed precision
                with torch.cuda.amp.autocast():
                    outputs = model(images)['out']
                    loss = criterion(outputs, masks)
                
                val_loss += loss.item()
                
                # Tính toán accuracy
                _, preds = torch.max(outputs, 1)
                valid_mask = (masks != 255)  # Bỏ qua ignore index
                correct_pixels += torch.sum(preds[valid_mask] == masks[valid_mask]).item()
                total_pixels += torch.sum(valid_mask).item()
        
        avg_val_loss = val_loss / len(val_loader)
        val_losses.append(avg_val_loss)
        
        # Tính mean IoU
        val_accuracy = correct_pixels / total_pixels if total_pixels > 0 else 0
        val_mious.append(val_accuracy)
        
        print(f'Epoch {epoch+1}/{num_epochs}: '
              f'Train Loss: {avg_train_loss:.4f}, '
              f'Val Loss: {avg_val_loss:.4f}, '
              f'Val mIoU: {val_accuracy:.4f}')
        
        # Lưu model tốt nhất dựa trên val loss
        if avg_val_loss < best_val_loss:
            best_val_loss = avg_val_loss
            torch.save(model.state_dict(), 'best_mobilenetv3.pth')
            print(f'New best model saved with val loss: {best_val_loss:.4f}')
            early_stop_counter = 0
        else:
            early_stop_counter += 1
        
        # Lưu model tốt nhất dựa trên mIoU
        if val_accuracy > best_miou:
            best_miou = val_accuracy
            torch.save(model.state_dict(), 'best_miou_mobilenetv3.pth')
            print(f'New best mIoU model saved with mIoU: {best_miou:.4f}')
        
        # Early stopping
        if early_stop_counter >= patience:
            print(f'Early stopping at epoch {epoch+1}')
            break
    
    # Plot training curves
    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    plt.plot(train_losses, label='Training Loss')
    plt.plot(val_losses, label='Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    
    plt.subplot(1, 2, 2)
    plt.plot(val_mious, label='Validation mIoU', color='green')
    plt.xlabel('Epoch')
    plt.ylabel('mIoU')
    plt.legend()
    
    plt.suptitle('Training Metrics')
    plt.savefig('bieudo_mobilenetv3.png')
    plt.close()
    
    print('Training complete')
    print(f'Best Validation Loss: {best_val_loss:.4f}')
    print(f'Best Validation mIoU: {best_miou:.4f}')
    
    return model
```

```python
# Check for GPU
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Using device: {device}')
    
# Initialize model
model = get_model(num_classes=3)
```

```python
# Train the model
trained_model = train_model(
    model, 
    train_loader, 
    val_loader, 
    num_epochs=30, 
    device=device
)
```

```python
def evaluate_full_metrics(model, dataset, num_classes=3, max_fps=10, device='cuda', num_samples=200):
    model.eval()
    loader = DataLoader(dataset, batch_size=1, shuffle=False)

    import time
    start = time.time()
    with torch.no_grad():
        for i, (image, _) in enumerate(loader):
            if i >= num_samples: break
            _ = model(image.to(device))['out']
    fps = num_samples / (time.time() - start)
    norm_fps = min(fps, max_fps) / max_fps

    y_true, y_pred = [], []
    with torch.no_grad():
        for i, (image, mask) in enumerate(loader):
            if i >= num_samples: break
            out = model(image.to(device))['out']
            pred = torch.argmax(out, dim=1).cpu().view(-1)
            mask = mask.view(-1)
            valid = (mask != 255)
            y_true.extend(mask[valid].numpy())
            y_pred.extend(pred[valid].numpy())
    p = precision_score(y_true, y_pred, average='macro', zero_division=0)
    r = recall_score(y_true, y_pred, average='macro', zero_division=0)
    f1 = 2 * p * r / (p + r + 1e-8)

    hist = np.zeros((num_classes, num_classes), dtype=np.int64)
    with torch.no_grad():
        for i, (image, mask) in enumerate(loader):
            if i >= num_samples: break
            out = model(image.to(device))['out']
            pred = torch.argmax(out, dim=1).cpu().squeeze()
            mask = mask.squeeze()
            valid = (mask != 255)
            for t, p in zip(mask[valid].view(-1), pred[valid].view(-1)):
                if t < num_classes:
                    hist[t, p] += 1
    iou = np.diag(hist) / (hist.sum(1) + hist.sum(0) - np.diag(hist) + 1e-8)
    miou = np.mean(iou)

    metric = 2 * norm_fps * f1 / (norm_fps + f1 + 1e-8)
    # -------IN KẾT QUẢ--------
    print(f"-     FPS: {fps:.2f}")
    print(f"- NormFPS: {norm_fps:.4f}")
    print(f"-      F1: {f1:.4f}")
    print(f"-    mIoU: {miou:.4f}")
    print(f"-  Metric: {metric:.4f}")
    return {"fps": fps, "f1": f1, "miou": miou, "metric": metric}
```

```python
evaluate_full_metrics(model, val_dataset, num_classes=3, device=device)
```

```python
def visualize_prediction(model, dataset, idx=0, device='cuda'):
    model.eval()
    image, mask = dataset[idx]
    image_tensor = image.unsqueeze(0).to(device)
    with torch.no_grad():
        output = model(image_tensor)['out']
        pred = torch.argmax(output, dim=1).squeeze().cpu().numpy()

    # ✅ Unnormalize ảnh
    img = image.permute(1, 2, 0).cpu().numpy()
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    img = (img * std + mean) * 255
    img = img.astype(np.uint8)
    
    gt = mask.cpu().numpy()

    def colorize(mask):
        colors = {
            0: (247, 195, 37),    # obstacle
            1: (41, 167, 224),    # water
            2: (90, 75, 164),     # sky
            255: (127, 127, 127)  # gray for ignored
        }
        out = np.zeros((*mask.shape, 3), dtype=np.uint8)
        for k, v in colors.items():
            out[mask == k] = v
        return out

    plt.figure(figsize=(12, 4))
    plt.subplot(1, 3, 1); plt.imshow(img.astype(np.uint8)); plt.title("Original Photo")
    plt.subplot(1, 3, 2); plt.imshow(colorize(gt)); plt.title("Ground Truth")
    plt.subplot(1, 3, 3); plt.imshow(colorize(pred)); plt.title("Predict")
    plt.show()
```

```python
visualize_prediction(model, val_dataset, idx=33)
```

```python
# Cell 3: Representative dataset từ tập val
import os
import numpy as np
from PIL import Image
import torchvision.transforms as T
import tensorflow as tf

def representative_data_gen():
    image_dir = "data/lars/images/val/images"
    image_files = [os.path.join(image_dir, f) for f in os.listdir(image_dir) if f.endswith(".jpg")]

    preprocess = T.Compose([
        T.Resize((320, 320)),
        T.ToTensor(),
        T.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225))
    ])

    for path in image_files[:100]:
        img = Image.open(path).convert("RGB")
        img = preprocess(img)
        img = img.unsqueeze(0).numpy().astype(np.float32)
        yield [img]
```

```python
# Cell 4: Convert SavedModel → .tflite INT8
converter = tf.lite.TFLiteConverter.from_saved_model("model_saved")
converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.representative_dataset = representative_data_gen
converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
converter.inference_input_type = tf.uint8
converter.inference_output_type = tf.uint8

int8_tflite_model = converter.convert()

with open("lraspp_mobilenetv3_int8.tflite", "wb") as f:
    f.write(int8_tflite_model)

print("✅ Đã tạo lraspp_mobilenetv3_int8.tflite")
```

```python
import torch
import torch.nn as nn
from torchvision.models.segmentation import lraspp_mobilenet_v3_large
from torch.utils.data import DataLoader
import albumentations as A
from albumentations.pytorch import ToTensorV2
import numpy as np
from PIL import Image
import os

# Load model đã train
model = lraspp_mobilenet_v3_large(pretrained=False)
model.classifier.low_classifier = nn.Conv2d(40, 3, kernel_size=1)
model.classifier.high_classifier = nn.Conv2d(128, 3, kernel_size=1)
model.load_state_dict(torch.load("mobilenetv3_pruned.pth", map_location='cpu'))

# Chuẩn bị Dataset (sử dụng lại class LaRSDataset của bạn)
train_dataset = LaRSDataset("data/lars", split="train", transform=transform)
val_dataset = LaRSDataset("data/lars", split="val", transform=transform)
train_loader = DataLoader(train_dataset, batch_size=8, shuffle=True)
```

```python
import torch.optim as optim
from tqdm import tqdm

# 1. Đóng băng backbone (chỉ train head)
for param in model.backbone.parameters():
    param.requires_grad = False

# 2. Optimizer và scheduler
optimizer = optim.AdamW([
    {'params': model.classifier.parameters(), 'lr': 1e-4},
    {'params': model.backbone.parameters(), 'lr': 1e-5}  # Mở băng một phần
], weight_decay=1e-5)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=10)

# 3. Hàm mất mát
criterion = nn.CrossEntropyLoss(ignore_index=255)

# 4. Fine-tuning trong 10 epochs
device = 'cuda' if torch.cuda.is_available() else 'cpu'
model.to(device)
best_loss = 1.0
for epoch in range(10):
    model.train()
    for images, masks in tqdm(train_loader):
        images, masks = images.to(device), masks.to(device)
        
        optimizer.zero_grad()
        outputs = model(images)['out']
        loss = criterion(outputs, masks)
        loss.backward()
        optimizer.step()
    
    scheduler.step()
    print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")
    if loss < best_loss:
        best_loss = loss
        torch.save(model.state_dict(), "mobilenetv3_finetuned.pth")
```

```python
import torch_pruning as tp

# 1. Load model đã fine-tuned
model.load_state_dict(torch.load("mobilenetv3_finetuned.pth"))
model.to(device)

# 2. Thiết lập pruner
pruner = tp.pruner.MagnitudePruner(
    model,
    example_inputs=torch.randn(1, 3, 320, 320).to(device),
    importance=tp.importance.MagnitudeImportance(p=1),  # L1-norm
    ch_sparsity=0.4,  # Prune 40%
    root_module_types=[nn.Conv2d],
    ignored_layers=[model.classifier]  # Giữ nguyên head
)

# 3. Thực hiện pruning
pruner.step()

# 4. Kiểm tra
print(f"Số tham số sau pruning: {sum(p.numel() for p in model.parameters()):,}")

# 5. Fine-tune ngắn 3 epochs
optimizer = optim.Adam(model.parameters(), lr=1e-5)
best_loss = 1.0
for epoch in range(3):
    model.train()
    for images, masks in train_loader:
        images, masks = images.to(device), masks.to(device)
        optimizer.zero_grad()
        outputs = model(images)['out']
        loss = criterion(outputs, masks)
        loss.backward()
        optimizer.step()
    print(f"Fine-tune sau pruning - Epoch {epoch+1}, Loss: {loss.item():.4f}")
    if loss < best_loss:
        best_loss = loss
        torch.save(model.state_dict(), "mobilenetv3_prunedv2.pth")
```
