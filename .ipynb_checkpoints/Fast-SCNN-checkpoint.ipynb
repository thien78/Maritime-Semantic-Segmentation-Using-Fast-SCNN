```python
import os
import numpy as np
from PIL import Image
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torch.nn.functional as F
import albumentations as A
from albumentations.pytorch import ToTensorV2
from albumentations import Compose, Resize, Normalize
import matplotlib.pyplot as plt
from sklearn.metrics import precision_score, recall_score
from tqdm import tqdm
```

```python
class LaRSDataset(Dataset):
    def __init__(self, root_dir, split='train', transform=None):
        self.image_dir = os.path.join(root_dir, 'images', split, 'images')
        self.mask_dir = os.path.join(root_dir, 'annotations', 'semantic', split, 'semantic_masks')
        self.image_files = sorted([f for f in os.listdir(self.image_dir) if f.endswith('.jpg')])
        self.transform = transform

    def __len__(self):
        return len(self.image_files)

    def __getitem__(self, idx):
        img_name = self.image_files[idx]
        image_path = os.path.join(self.image_dir, img_name)
        mask_path = os.path.join(self.mask_dir, img_name.replace('.jpg', '.png'))

        image = np.array(Image.open(image_path).convert('RGB'))
        mask = np.array(Image.open(mask_path)).astype(np.uint8)
        mask[mask > 2] = 255

        if self.transform:
            transformed = self.transform(image=image, mask=mask)
            image = transformed['image']
            mask = transformed['mask']

        return image, mask.long()
```

```python
# Cấu hình transform dùng chung
transform = Compose([
    Resize(320, 320),
    Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
    ToTensorV2()
])

# Dataset
train_dataset = LaRSDataset("data/lars", split="train", transform=transform)
val_dataset   = LaRSDataset("data/lars", split="val", transform=transform)

# DataLoader
train_loader = DataLoader(train_dataset, batch_size=4, shuffle=True, num_workers=0, drop_last=True)
val_loader   = DataLoader(val_dataset, batch_size=4, shuffle=False, num_workers=0, drop_last=True)
```

```python
class FastSCNN(nn.Module):
    def __init__(self, num_classes):
        super(FastSCNN, self).__init__()
        
        # Learning to Downsample module
        self.lds = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, stride=2, padding=1, bias=False),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.Conv2d(32, 48, kernel_size=3, stride=2, padding=1, bias=False),
            nn.BatchNorm2d(48),
            nn.ReLU(),
            nn.Conv2d(48, 64, kernel_size=3, stride=2, padding=1, bias=False),
            nn.BatchNorm2d(64),
            nn.ReLU()
        )
        
        # Global Feature Extractor
        self.gfe = nn.Sequential(
            MobileNetV2Block(64, 64, expand_ratio=6, stride=2),
            MobileNetV2Block(64, 64, expand_ratio=6, stride=1),
            MobileNetV2Block(64, 64, expand_ratio=6, stride=1),
            MobileNetV2Block(64, 96, expand_ratio=6, stride=2),
            MobileNetV2Block(96, 96, expand_ratio=6, stride=1),
            MobileNetV2Block(96, 96, expand_ratio=6, stride=1),
            MobileNetV2Block(96, 128, expand_ratio=6, stride=1),
            MobileNetV2Block(128, 128, expand_ratio=6, stride=1),
            MobileNetV2Block(128, 128, expand_ratio=6, stride=1)
        )
        
        # Feature Fusion
        self.ffm = FeatureFusionModule(64, 128, 128)
        
        # Classifier
        self.classifier = nn.Sequential(
            nn.Dropout2d(0.1),
            nn.Conv2d(128, num_classes, kernel_size=1)
        )
        
        # Auxiliary classifier
        self.aux = nn.Sequential(
            nn.Conv2d(64, 32, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.Dropout2d(0.1),
            nn.Conv2d(32, num_classes, kernel_size=1)
        )
    
    def forward(self, x):
        size = x.size()[2:]
        
        # Learning to Downsample
        x_lds = self.lds(x)
        
        # Global Feature Extractor
        x_gfe = self.gfe(x_lds)
        
        # Feature Fusion
        x_ffm = self.ffm(x_lds, x_gfe)
        
        # Classifier
        x_out = self.classifier(x_ffm)
        x_out = F.interpolate(x_out, size=size, mode='bilinear', align_corners=True)
        
        # Auxiliary output
        x_aux = self.aux(x_lds)
        x_aux = F.interpolate(x_aux, size=size, mode='bilinear', align_corners=True)
        
        return x_out, x_aux


class MobileNetV2Block(nn.Module):
    def __init__(self, in_channels, out_channels, expand_ratio, stride):
        super(MobileNetV2Block, self).__init__()
        self.stride = stride
        hidden_dim = in_channels * expand_ratio
        
        self.use_res_connect = self.stride == 1 and in_channels == out_channels
        
        if expand_ratio == 1:
            self.conv = nn.Sequential(
                nn.Conv2d(hidden_dim, hidden_dim, 3, stride, 1, groups=hidden_dim, bias=False),
                nn.BatchNorm2d(hidden_dim),
                nn.ReLU6(),
                nn.Conv2d(hidden_dim, out_channels, 1, 1, 0, bias=False),
                nn.BatchNorm2d(out_channels)
            )
        else:
            self.conv = nn.Sequential(
                nn.Conv2d(in_channels, hidden_dim, 1, 1, 0, bias=False),
                nn.BatchNorm2d(hidden_dim),
                nn.ReLU6(),
                nn.Conv2d(hidden_dim, hidden_dim, 3, stride, 1, groups=hidden_dim, bias=False),
                nn.BatchNorm2d(hidden_dim),
                nn.ReLU6(),
                nn.Conv2d(hidden_dim, out_channels, 1, 1, 0, bias=False),
                nn.BatchNorm2d(out_channels)
            )

    def forward(self, x):
        if self.use_res_connect:
            return x + self.conv(x)
        else:
            return self.conv(x)


class FeatureFusionModule(nn.Module):
    def __init__(self, highter_in_channels, lower_in_channels, out_channels, scale_factor=4):
        super(FeatureFusionModule, self).__init__()
        self.scale_factor = scale_factor
        
        self.dwconv = nn.Sequential(
            nn.Conv2d(lower_in_channels, out_channels, kernel_size=3, stride=1, padding=1, groups=out_channels, bias=False),
            nn.BatchNorm2d(out_channels)
        )
        
        self.conv_lower_res = nn.Sequential(
            nn.Conv2d(out_channels, out_channels, kernel_size=1),
            nn.BatchNorm2d(out_channels)
        )
        
        self.conv_higher_res = nn.Sequential(
            nn.Conv2d(highter_in_channels, out_channels, kernel_size=1),
            nn.BatchNorm2d(out_channels)
        )
        
        self.relu = nn.ReLU()

    def forward(self, higher_res_feature, lower_res_feature):
        lower_res_feature = F.interpolate(lower_res_feature, scale_factor=self.scale_factor, mode='bilinear', align_corners=True)
        lower_res_feature = self.dwconv(lower_res_feature)
        lower_res_feature = self.conv_lower_res(lower_res_feature)
        
        higher_res_feature = self.conv_higher_res(higher_res_feature)
        
        out = higher_res_feature + lower_res_feature
        return self.relu(out)
```

```python
class ConfusionMatrix:
    def __init__(self, num_classes):
        self.num_classes = num_classes
        self.mat = None

    def update(self, preds, targets):
        n = self.num_classes
        if self.mat is None:
            self.mat = torch.zeros((n, n), dtype=torch.int64, device=preds.device)
        with torch.no_grad():
            k = (targets >= 0) & (targets < n)
            inds = n * targets[k].to(torch.int64) + preds[k]
            self.mat += torch.bincount(inds, minlength=n**2).reshape(n, n)

    def compute_miou(self):
        h = self.mat.float()
        iou = torch.diag(h) / (h.sum(1) + h.sum(0) - torch.diag(h) + 1e-6)
        return iou.mean().item()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.cuda.amp import GradScaler, autocast
from torch.optim.swa_utils import AveragedModel, SWALR
from torchmetrics import JaccardIndex
from tqdm import tqdm

def train_model(model, train_loader, val_loader, num_epochs=50, lr=0.001, warmup_epochs=5, accumulation_steps=4):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model = model.to(device)
    
    # 1. Khởi tạo optimizer và scheduler
    criterion = nn.CrossEntropyLoss(ignore_index=255)
    optimizer = optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.OneCycleLR(
        optimizer, 
        max_lr=lr, 
        epochs=num_epochs, 
        steps_per_epoch=len(train_loader)//accumulation_steps
    )
    
    # 2. SWA (Stochastic Weight Averaging)
    swa_model = torch.optim.swa_utils.AveragedModel(model)
    swa_start = int(num_epochs * 0.75)
    swa_scheduler = torch.optim.swa_utils.SWALR(optimizer, swa_lr=lr*0.1)
    
    # 3. AMP (Automatic Mixed Precision)
    scaler = GradScaler()
    best_miou = 0.0
    patience = 5
    epochs_no_improve = 0

    for epoch in range(num_epochs):
        model.train()
        train_loss = 0.0
        optimizer.zero_grad()
        
        # 4. Learning Rate Warmup
        if epoch < warmup_epochs:
            for param_group in optimizer.param_groups:
                param_group['lr'] = lr * ((epoch + 1) / warmup_epochs)

        # 5. Training với Gradient Accumulation
        train_bar = tqdm(train_loader, desc=f'Epoch {epoch+1}/{num_epochs}')
        for i, (images, masks) in enumerate(train_bar):
            images, masks = images.to(device), masks.to(device)
            
            with autocast():
                outputs, aux_outputs = model(images)
                loss = (criterion(outputs, masks) + 0.4 * criterion(aux_outputs, masks)) / accumulation_steps
            
            scaler.scale(loss).backward()
            
            if (i + 1) % accumulation_steps == 0:
                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad()
                
                # 6. Cập nhật learning rate
                if epoch < swa_start:
                    scheduler.step()
                else:
                    swa_scheduler.step()
            
            train_loss += loss.item() * accumulation_steps
            train_bar.set_postfix(loss=train_loss/(i+1))

        # 7. Validation với ConfusionMatrix
        model.eval()
        confmat = ConfusionMatrix(num_classes=3)
        val_loss = 0.0
        
        with torch.no_grad():
            val_bar = tqdm(val_loader, desc='Validation')
            for images, masks in val_bar:
                images, masks = images.to(device), masks.to(device)
                outputs, _ = model(images)
                loss = criterion(outputs, masks)
                val_loss += loss.item()
                confmat.update(outputs.argmax(1), masks)
        
        miou = confmat.compute_miou()
        print(f'Epoch {epoch+1}: Train Loss = {train_loss/len(train_loader):.4f}, '
              f'Val Loss = {val_loss/len(val_loader):.4f}, '
              f'mIoU = {miou:.4f}')

        # 8. Early Stopping
        if miou > best_miou:
            best_miou = miou
            epochs_no_improve = 0
            torch.save(model.state_dict(), 'best_model.pth')
        else:
            epochs_no_improve += 1
            if epochs_no_improve == patience:
                print(f'Early stopping at epoch {epoch+1}!')
                break
        
        # 9. Cập nhật SWA model
        if epoch >= swa_start:
            swa_model.update_parameters(model)

    # 10. Áp dụng SWA cuối cùng
    if num_epochs > swa_start:
        torch.optim.swa_utils.update_bn(train_loader, swa_model, device=device)
        torch.save(swa_model.state_dict(), 'swa_model.pth')
        return swa_model
    
    return model
```

```python
# Initialize model and train
model = FastSCNN(num_classes=3)  # 3 classes (0, 1, 2 and 255 is ignored)
```

```python
optimized_model = train_model(
    model, 
    train_loader, 
    val_loader,
    num_epochs=30,
    lr=0.001,
    warmup_epochs=5,
    accumulation_steps=4
)
```

```python
# Cell 3: Chuẩn bị representative dataset
import numpy as np
import tensorflow as tf
from PIL import Image
import os

def representative_data_gen():
    image_dir = "data/lars/images/val/images"
    image_files = [os.path.join(image_dir, f) for f in os.listdir(image_dir) if f.endswith(".jpg")]

    preprocess = T.Compose([
        T.Resize((320, 320)),
        T.ToTensor(),
        T.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225))
    ])

    for image_path in image_files[:100]:  # lấy 100 ảnh để làm mẫu
        img = Image.open(image_path).convert("RGB")
        img = preprocess(img)
        img = img.numpy()
        img = np.expand_dims(img, axis=0)
        img = img.astype(np.float32)
        yield [img]
```

```python
def evaluate_full_metrics(model, dataset, num_classes=3, max_fps=10, device='cuda', num_samples=200):
    model.eval()
    loader = DataLoader(dataset, batch_size=1, shuffle=False)

    import time
    start = time.time()
    with torch.no_grad():
        for i, (image, _) in enumerate(loader):
            if i >= num_samples: break
            _ = model(image.to(device))[0]#['out']
    fps = num_samples / (time.time() - start)
    norm_fps = min(fps, max_fps) / max_fps

    y_true, y_pred = [], []
    with torch.no_grad():
        for i, (image, mask) in enumerate(loader):
            if i >= num_samples: break
            out = model(image.to(device))[0]#['out']
            pred = torch.argmax(out, dim=1).cpu().view(-1)
            mask = mask.view(-1)
            valid = (mask != 255)
            y_true.extend(mask[valid].numpy())
            y_pred.extend(pred[valid].numpy())
    p = precision_score(y_true, y_pred, average='macro', zero_division=0)
    r = recall_score(y_true, y_pred, average='macro', zero_division=0)
    f1 = 2 * p * r / (p + r + 1e-8)

    hist = np.zeros((num_classes, num_classes), dtype=np.int64)
    with torch.no_grad():
        for i, (image, mask) in enumerate(loader):
            if i >= num_samples: break
            out = model(image.to(device))[0]#['out']
            pred = torch.argmax(out, dim=1).cpu().squeeze()
            mask = mask.squeeze()
            valid = (mask != 255)
            for t, p in zip(mask[valid].view(-1), pred[valid].view(-1)):
                if t < num_classes:
                    hist[t, p] += 1
    iou = np.diag(hist) / (hist.sum(1) + hist.sum(0) - np.diag(hist) + 1e-8)
    miou = np.mean(iou)

    metric = 2 * norm_fps * f1 / (norm_fps + f1 + 1e-8)
    # -------IN KẾT QUẢ--------
    print(f"-     FPS: {fps:.2f}")
    print(f"- NormFPS: {norm_fps:.4f}")
    print(f"-      F1: {f1:.4f}")
    print(f"-    mIoU: {miou:.4f}")
    print(f"-  Metric: {metric:.4f}")
    return {"fps": fps, "f1": f1, "miou": miou, "metric": metric}
```

```python
evaluate_full_metrics(model, val_dataset, num_classes=3, device=device)
```

```python
def visualize_prediction(model, dataset, idx=0, device='cuda'):
    model.eval()
    image, mask = dataset[idx]
    image_tensor = image.unsqueeze(0).to(device)
    with torch.no_grad():
        output = model(image_tensor)[0]#['out']
        pred = torch.argmax(output, dim=1).squeeze().cpu().numpy()

    # ✅ Unnormalize ảnh
    img = image.permute(1, 2, 0).cpu().numpy()
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    img = (img * std + mean) * 255
    img = img.astype(np.uint8)
    
    gt = mask.cpu().numpy()

    def colorize(mask):
        colors = {
            0: (247, 195, 37),    # obstacle
            1: (41, 167, 224),    # water
            2: (90, 75, 164),     # sky
            255: (127, 127, 127)  # gray for ignored
        }
        out = np.zeros((*mask.shape, 3), dtype=np.uint8)
        for k, v in colors.items():
            out[mask == k] = v
        return out

    plt.figure(figsize=(12, 4))
    plt.subplot(1, 3, 1); plt.imshow(img.astype(np.uint8)); plt.title("Original Photo")
    plt.subplot(1, 3, 2); plt.imshow(colorize(gt)); plt.title("Ground Truth")
    plt.subplot(1, 3, 3); plt.imshow(colorize(pred)); plt.title("Predict")
    plt.show()
```

```python
visualize_prediction(model, val_dataset, idx=131)
```
