```python
# CELL1: Import các thư viện cần thiết (cập nhật)
import os
import numpy as np
from PIL import Image
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import albumentations as A
from albumentations.pytorch import ToTensorV2
from tqdm import tqdm
import torch.cuda.amp as amp
import time
from torch.nn.parallel import DistributedDataParallel as DDP
import torch.distributed as dist
```

```python
# ✅ Dataset xử lý Semantic Segmentation từ LaRS
class LaRSDataset(Dataset):
    def __init__(self, root_dir, split='train', transform=None):
        self.image_dir = os.path.join(root_dir, 'images', split, 'images')
        self.mask_dir = os.path.join(root_dir, 'annotations', 'semantic', split, 'semantic_masks')
        self.image_files = sorted([f for f in os.listdir(self.image_dir) if f.endswith('.jpg')])
        self.transform = transform

    def __len__(self):
        return len(self.image_files)

    def __getitem__(self, idx):
        img_name = self.image_files[idx]
        image_path = os.path.join(self.image_dir, img_name)
        mask_path = os.path.join(self.mask_dir, img_name.replace('.jpg', '.png'))

        image = np.array(Image.open(image_path).convert('RGB'))
        mask = np.array(Image.open(mask_path)).astype(np.uint8)
        mask[mask > 2] = 255

        if self.transform:
            transformed = self.transform(image=image, mask=mask)
            image = transformed['image']
            mask = transformed['mask']

        return image, mask.long()
```

```python
# CELL3: Cập nhật transform với augmentation mạnh hơn
# Tăng cường augmentation cho tập train
train_transform = A.Compose([
    A.Resize(320, 320),
    A.HorizontalFlip(p=0.5),
    A.VerticalFlip(p=0.1),
    A.RandomRotate90(p=0.3),
    A.ShiftScaleRotate(shift_limit=0.1, scale_limit=0.1, rotate_limit=15, p=0.5),
    A.RandomBrightnessContrast(brightness_limit=0.2, contrast_limit=0.2, p=0.5),
    A.OneOf([
        A.GaussianBlur(blur_limit=(3, 5)),
        A.MotionBlur(blur_limit=5),
        A.MedianBlur(blur_limit=3)
    ], p=0.3),
    A.GridDropout(ratio=0.2, random_offset=True, holes_number_x=4, holes_number_y=4, p=0.2),
    A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
    ToTensorV2()
])

# Transform đơn giản cho validation
val_transform = A.Compose([
    A.Resize(320, 320),
    A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
    ToTensorV2()
])

# Dataset
train_dataset = LaRSDataset("data/lars", split="train", transform=train_transform)
val_dataset = LaRSDataset("data/lars", split="val", transform=val_transform)

# DataLoader - Tăng batch size và sử dụng prefetch
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True, 
                          num_workers=0, pin_memory=True, drop_last=True)
val_loader = DataLoader(val_dataset, batch_size=16, shuffle=False, 
                        num_workers=0, pin_memory=True, drop_last=True)
```

```python
# CELL4: Tối ưu kiến trúc STDC-Seg với Depth-wise Separable Conv và Attention
class DepthwiseSeparableConv(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_size=3, stride=1, padding=1):
        super().__init__()
        self.depthwise = nn.Conv2d(in_channels, in_channels, kernel_size=kernel_size, 
                                  stride=stride, padding=padding, groups=in_channels)
        self.pointwise = nn.Conv2d(in_channels, out_channels, kernel_size=1)
        self.bn = nn.BatchNorm2d(out_channels)
        self.relu = nn.ReLU(inplace=True)
        
    def forward(self, x):
        x = self.depthwise(x)
        x = self.pointwise(x)
        x = self.bn(x)
        return self.relu(x)

class ChannelAttention(nn.Module):
    def __init__(self, in_channels, reduction_ratio=8):
        super().__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.max_pool = nn.AdaptiveMaxPool2d(1)
        
        self.fc = nn.Sequential(
            nn.Linear(in_channels, in_channels // reduction_ratio),
            nn.ReLU(inplace=True),
            nn.Linear(in_channels // reduction_ratio, in_channels),
            nn.Sigmoid()
        )
        
    def forward(self, x):
        b, c, _, _ = x.size()
        avg_out = self.fc(self.avg_pool(x).view(b, c))
        max_out = self.fc(self.max_pool(x).view(b, c))
        out = avg_out + max_out
        return out.view(b, c, 1, 1)

class SpatialAttention(nn.Module):
    def __init__(self, kernel_size=7):
        super().__init__()
        self.conv = nn.Conv2d(2, 1, kernel_size=kernel_size, padding=kernel_size//2)
        self.sigmoid = nn.Sigmoid()
        
    def forward(self, x):
        avg_out = torch.mean(x, dim=1, keepdim=True)
        max_out, _ = torch.max(x, dim=1, keepdim=True)
        x = torch.cat([avg_out, max_out], dim=1)
        x = self.conv(x)
        return self.sigmoid(x)

class EfficientAttentionBlock(nn.Module):
    """Light-weight attention block tối ưu cho real-time"""
    def __init__(self, in_channels):
        super().__init__()
        self.channel_att = ChannelAttention(in_channels)
        self.spatial_att = SpatialAttention()
        
    def forward(self, x):
        channel_att = self.channel_att(x)
        spatial_att = self.spatial_att(x)
        
        # Kết hợp attention
        att = channel_att * spatial_att
        return x * att + x

class STDCModule(nn.Module):
    def __init__(self, in_chan, out_chan, stride=1, use_attention=False):
        super().__init__()
        # Sử dụng depth-wise separable conv để giảm tham số
        self.conv1 = DepthwiseSeparableConv(in_chan, out_chan//2, stride=1)
        self.conv2 = DepthwiseSeparableConv(out_chan//2, out_chan//2, stride=stride)
        self.conv3 = DepthwiseSeparableConv(out_chan//2, out_chan, stride=1)
        
        self.shortcut = nn.Sequential()
        if stride != 1 or in_chan != out_chan:
            self.shortcut = nn.Sequential(
                DepthwiseSeparableConv(in_chan, out_chan, stride=stride)
            )
            
        self.attention = EfficientAttentionBlock(out_chan) if use_attention else None
        
    def forward(self, x):
        out = self.conv1(x)
        out = self.conv2(out)
        out = self.conv3(out)
        out = out + self.shortcut(x)
        
        if self.attention:
            out = self.attention(out)
            
        return out

class STDCNet(nn.Module):
    def __init__(self, num_classes=3):
        super().__init__()
        # Stage 1 - Giảm kích thước đầu vào
        self.conv1 = DepthwiseSeparableConv(3, 32, stride=2)
        self.conv2 = DepthwiseSeparableConv(32, 64, stride=2)
        
        # Stage 2-5 với attention ở các giai đoạn sau
        self.stage2 = nn.Sequential(
            STDCModule(64, 128, stride=2),
            STDCModule(128, 128, stride=1, use_attention=True)
        )
        self.stage3 = nn.Sequential(
            STDCModule(128, 256, stride=2),
            STDCModule(256, 256, stride=1, use_attention=True)
        )
        self.stage4 = nn.Sequential(
            STDCModule(256, 512, stride=2),
            STDCModule(512, 512, stride=1)
        )
        self.stage5 = nn.Sequential(
            STDCModule(512, 1024, stride=2),
            STDCModule(1024, 1024, stride=1)
        )
        
        # Context Module - Thu thập thông tin đa tầng
        self.context = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Conv2d(1024, 256, 1),
            nn.ReLU(inplace=True),
            nn.Upsample(size=(5, 5), mode='bilinear', align_corners=True)
        )
        
        # Segmentation head hiệu quả
        self.seg_head = nn.Sequential(
            DepthwiseSeparableConv(1024 + 256, 256, kernel_size=3, padding=1),
            EfficientAttentionBlock(256),
            nn.Conv2d(256, num_classes, kernel_size=1)
        )
        
    def forward(self, x):
        # Downsample 1/4
        x = self.conv1(x)
        x = self.conv2(x)
        
        # Downsample 1/8
        s2 = self.stage2(x)
        
        # Downsample 1/16
        s3 = self.stage3(s2)
        
        # Downsample 1/32
        s4 = self.stage4(s3)
        
        # Downsample 1/64
        s5 = self.stage5(s4)
        
        # Context module
        context = self.context(s5)
        
        # Kết hợp feature map và context
        s5 = torch.cat([s5, context], dim=1)
        
        # Segmentation head
        out = self.seg_head(s5)
        out = F.interpolate(out, size=(320, 320), mode='bilinear', align_corners=True)
        return out
```

```python
# CELL5: Hàm train_model tối ưu với Mixed Precision và Asymmetric Loss
class AsymmetricLoss(nn.Module):
    """Loss function giải quyết class imbalance"""
    def __init__(self, gamma_neg=2, gamma_pos=0, eps=1e-8, ignore_index=255):
        super().__init__()
        self.gamma_neg = gamma_neg
        self.gamma_pos = gamma_pos
        self.eps = eps
        self.ignore_index = ignore_index
        
    def forward(self, logits, targets):
        # Chuyển đổi logits từ [B, C, H, W] -> [B, H, W, C] để xử lý mask
        logits = logits.permute(0, 2, 3, 1)  # [B, H, W, C]
        
        # Tạo mask và lọc các pixel hợp lệ
        valid_mask = targets != self.ignore_index
        logits = logits[valid_mask]  # [N_valid, C]
        targets = targets[valid_mask]  # [N_valid]
        
        # Nếu không có pixel hợp lệ, trả về loss = 0
        if targets.numel() == 0:
            return torch.tensor(0.0, device=logits.device, requires_grad=True)
        
        # Tính toán loss
        logpt = F.log_softmax(logits, dim=-1)
        logpt = logpt.gather(1, targets.unsqueeze(1)).squeeze()
        pt = logpt.exp()
        
        # Áp dụng asymmetric focusing
        loss = -((1 - pt).pow(self.gamma_neg) * logpt).mean()
        return loss

def train_model(model, train_loader, val_loader, num_epochs=25, lr=0.01):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model = model.to(device)
    
    # Sử dụng AsymmetricLoss để giải quyết class imbalance
    criterion = AsymmetricLoss(gamma_neg=2, gamma_pos=0.5)
    
    # Optimizer tối ưu với weight decay và momentum
    optimizer = optim.SGD(model.parameters(), lr=lr, momentum=0.9, weight_decay=1e-4)
    
    # Learning rate scheduler với warmup
    warmup_epochs = 3
    scheduler = optim.lr_scheduler.OneCycleLR(
        optimizer, 
        max_lr=lr, 
        steps_per_epoch=len(train_loader), 
        epochs=num_epochs,
        pct_start=warmup_epochs/num_epochs
    )
    
    best_val_loss = float('inf')
    scaler = amp.GradScaler()  # Cho mixed precision
    
    # Biến để theo dõi hiệu suất
    train_losses = []
    val_losses = []
    
    start_time = time.time()
    
    for epoch in range(num_epochs):
        model.train()
        epoch_train_loss = 0.0
        
        # Training phase với mixed precision
        for images, masks in tqdm(train_loader, desc=f'Epoch {epoch+1}/{num_epochs} - Training'):
            images = images.to(device, non_blocking=True)
            masks = masks.to(device, non_blocking=True)
            
            optimizer.zero_grad()
            
            # Mixed precision training
            with amp.autocast():
                outputs = model(images)
                loss = criterion(outputs, masks)
            
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
            scheduler.step()
            
            epoch_train_loss += loss.item() * images.size(0)
        
        # Tính toán train loss
        epoch_train_loss /= len(train_loader.dataset)
        train_losses.append(epoch_train_loss)
        
        # Validation phase
        model.eval()
        epoch_val_loss = 0.0
        with torch.no_grad():
            for images, masks in tqdm(val_loader, desc=f'Epoch {epoch+1}/{num_epochs} - Validation'):
                images = images.to(device, non_blocking=True)
                masks = masks.to(device, non_blocking=True)
                
                outputs = model(images)
                loss = criterion(outputs, masks)
                epoch_val_loss += loss.item() * images.size(0)
        
        epoch_val_loss /= len(val_loader.dataset)
        val_losses.append(epoch_val_loss)
        
        print(f'Epoch {epoch+1}/{num_epochs} | '
              f'Train Loss: {epoch_train_loss:.4f} | '
              f'Val Loss: {epoch_val_loss:.4f} | '
              f'LR: {scheduler.get_last_lr()[0]:.6f}')
        
        # Lưu model tốt nhất
        if epoch_val_loss < best_val_loss:
            best_val_loss = epoch_val_loss
            torch.save(model.state_dict(), 'best_stdcseg.pth')
            print('Model saved!')
    
    training_time = time.time() - start_time
    print(f'Training complete in {training_time//60:.0f}m {training_time%60:.0f}s')
    
    # In kết quả cuối cùng
    print(f'Best validation loss: {best_val_loss:.4f}')
    
    return model, train_losses, val_losses

# Khởi tạo và huấn luyện mô hình
model = STDCNet(num_classes=3)
print(f"Số lượng tham số: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}")

trained_model, train_losses, val_losses = train_model(
    model, train_loader, val_loader, 
    num_epochs=30, lr=0.02
)

# Đánh giá tốc độ inference
def benchmark_model(model, input_size=(1, 3, 320, 320), num_runs=100):
    device = next(model.parameters()).device
    input_tensor = torch.randn(input_size).to(device)
    
    # Warm-up
    for _ in range(10):
        _ = model(input_tensor)
    
    # Benchmark
    start_time = time.time()
    for _ in range(num_runs):
        _ = model(input_tensor)
    end_time = time.time()
    
    avg_time = (end_time - start_time) / num_runs
    fps = 1 / avg_time
    print(f"Average inference time: {avg_time*1000:.2f}ms")
    print(f"FPS: {fps:.2f}")
    return fps

model.load_state_dict(torch.load('best_stdcseg.pth'))
model.eval()
benchmark_model(model)
```

```python
import os
import numpy as np
from PIL import Image
import torchvision.transforms as T
import tensorflow as tf

def representative_data_gen():
    image_dir = "data/lars/images/val/images"
    image_files = [os.path.join(image_dir, f) for f in os.listdir(image_dir) if f.endswith(".jpg")]

    preprocess = T.Compose([
        T.Resize((320, 320)),
        T.ToTensor(),
        T.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225))
    ])

    for image_path in image_files[:100]:  # lấy 100 ảnh để làm mẫu
        img = Image.open(image_path).convert("RGB")
        img = preprocess(img)
        img = img.numpy()
        img = np.expand_dims(img, axis=0)
        img = img.astype(np.float32)
        yield [img]
```

```python
# Cell 4: Chuyển từ SavedModel sang TFLite int8
converter = tf.lite.TFLiteConverter.from_saved_model("model_saved")
converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.representative_dataset = representative_data_gen
converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
converter.inference_input_type = tf.uint8
converter.inference_output_type = tf.uint8

tflite_model = converter.convert()

with open("stdcseg_int8.tflite", "wb") as f:
    f.write(tflite_model)

print("✅ Đã tạo stdcseg_int8.tflite")
```

```python
import tensorflow as tf
import numpy as np

interpreter = tf.lite.Interpreter(model_path="stdcseg_int8.tflite")
interpreter.allocate_tensors()
tensor_details = interpreter.get_tensor_details()

# Lọc các tensor có tên chứa Conv hoặc depthwise, thường là trọng số
weight_like_tensors = [
    t for t in tensor_details
    if ('Conv' in t['name'] or 'depthwise' in t['name'])
    and np.prod(t['shape']) > 1
    and t['dtype'] in (np.int8, np.uint8)
]

total_weight_params = sum(np.prod(t['shape']) for t in weight_like_tensors)

print(f"Approx number of weight params: {total_weight_params:,}")
print("Sample tensor names:")
for t in weight_like_tensors[:10]:
    print(f"- {t['name']}, shape={t['shape']}")
```
